```python
from pathlib import Path
from collections import Counter

DATA_DIR = PROJECT_ROOT / "Data"

print("=" * 100)
print("DATA DIRECTORY EXPLORATION")
print("=" * 100)

for folder in sorted(DATA_DIR.rglob("*")):

    if not folder.is_dir():
        continue

    files = [
        f for f in folder.iterdir()
        if f.is_file()
    ]

    print("\n" + "=" * 100)
    print(f"Folder: {folder.relative_to(PROJECT_ROOT)}")
    print("=" * 100)

    print(f"Total Files: {len(files)}")

    if len(files) == 0:
        print("Empty Folder")
        continue

    extension_counter = Counter(
        f.suffix.lower()
        for f in files
    )

    print("\nFile Types:")

    for ext, count in sorted(extension_counter.items()):

        print(
            f"  {ext:<10} {count}"
        )

    print("\nSample Files:")

    for f in files[:5]:

        print(
            f"  {f.name}"
        )
```

```python
from pathlib import Path
from collections import Counter

DATA_DIR = PROJECT_ROOT / "Data"

extensions = Counter()

for file in DATA_DIR.rglob("*"):

    if file.is_file():

        ext = file.suffix.lower()

        if ext:
            extensions[ext] += 1

print("File Types")
print("=" * 30)

for ext, count in sorted(extensions.items()):

    print(
        f"{ext:<10} {count}"
    )
```

```python
import pandas as pd
from pathlib import Path

rows = []

for folder in (PROJECT_ROOT / "Data").rglob("*"):

    if folder.is_dir():

        files = [
            f for f in folder.iterdir()
            if f.is_file()
        ]

        rows.append({
            "folder": str(folder.relative_to(PROJECT_ROOT)),
            "file_count": len(files)
        })

df = pd.DataFrame(rows)

df.sort_values(
    "file_count",
    ascending=False
)
```
